```python
# 24. Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.

import time

n=1000000
py_list=list(range(1,n+1))
np_array = np.arange(1,n+1)

start_time1= time.time()
py_result = [x * 2 for x in py_list]
py_time = time.time() - start_time1

start_time = time.time()
np_result = np_array * 2
np_time = time.time() - start_time

print(f"Python list time: {py_time:.5f} seconds")
print(f"NumPy array time: {np_time:.5f} seconds")
print(f"NumPy is {py_time / np_time:.1f}x faster")
```
