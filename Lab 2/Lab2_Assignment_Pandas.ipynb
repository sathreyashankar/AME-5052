1. Create a Pandas Series containing the marks: 78, 85, 92, 67, 88

    Display:

	Complete Series
	Values
	Index
	Data type
	Mean
	Maximum
	Minimum

```python
marks=pd.Series([78, 85, 92, 67, 88])

print('Marks (Series): \n', marks)
print('Marks Values: ', marks.values)
print('Marks(Index): ', marks.index)
print('Marks DataType: ', marks.dtype)
print('Marks Mean: ', marks.mean())
print('Max Marks: ', marks.max())
print('Min Marks: ', marks.min())
```

3. Create the following DataFrame:

| Name  | Age | Department | Marks |
| ----- | --: | ---------- | ----: |
| Asha  |  22 | CSE        |    85 |
| Ravi  |  23 | ECE        |    78 |
| Meena |  21 | CSE        |    92 |
| Kiran |  24 | ME         |    74 |
| Arun  |  22 | ECE        |    88 |

Display:

	Complete DataFrame
	First 3 rows
	Last 2 rows
	Shape
	Column names
	Data types

```python
marks=pd.DataFrame({
    'Name':["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    'Marks':[85, 78, 92, 74, 88],
    'Age':[22,23,21,24,22],
    'Department':['CSE','ECE',"CSE","ME","ECE"]
})

print('Marks (DataFrame): \n', marks)
print('First Three Rows of Marks : \n', marks.head(3))
print('Last Two Rows of Marks : \n', marks.tail(2))
print('Shape of Marks : \n', marks.shape)
print('Column Names: ', marks.columns)
print('Data Types: ', marks.dtypes)

```

4. Using the previous DataFrame, display:

	- info()
	- describe()
	- Number of rows
	- Number of columns

```python
print('Marks info: ', marks.info())
print('Marks describe: ', marks.describe())
print('No of Rows: ', marks.shape[0])
print('No of Cols: ', marks.shape[1])
```

5. Using marks, display:

	Name column
	Marks column
	Name and Marks together

```python
print('Name column: ', marks.iloc[:,0])
print('Marks column: ', marks.iloc[:,-1])
print('Name and Marks together: ', marks.loc[:,['Name','Marks']])
```

6. Display:

	- First row using iloc
	- Third row using iloc

```python
print('First row using iloc: \n', marks.iloc[1,:])
print('\nThird row using iloc: \n', marks.iloc[3,:])
print('\nRows 1 to 3: \n', marks.iloc[0:2,:])
print('\nName and Marks columns using loc: \n', marks.loc[:,['Name','Marks']])
```

7. Find:

	- Students scoring above 80.
	- Students scoring between 75 and 90.
	- Students belonging to CSE.
	- CSE students scoring above 80.

```python
print('Students scoring above 80: \n', marks[marks['Marks']>80])
print('\nStudents scoring between 75 and 90: \n', marks[(marks['Marks']>75) & (marks['Marks']<90)])
print('\nStudents belonging to CSE: \n', marks[marks['Department']=='CSE'])
print('\nCSE students scoring above 80: \n', marks[(marks['Department']=='CSE') & (marks['Marks']>80)])
```

8. Display students belonging to either CSE or ECE.

```python
print('Display students belonging to either CSE or ECE: \n', marks[(marks['Department'] == 'CSE') | (marks['Department'] == 'ECE')])
```

9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".

```python
marks['Result'] = marks['Marks'].apply(lambda x: 'Pass' if x >= 75 else 'Fail')

print('DataFrame with Result column: \n', marks)
```


10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |

```python
marks['Grade'] = marks['Marks'].apply(
    lambda x: 'A' if x >= 90 else
              'B' if x >= 80 else
              'C' if x >= 70 else
              'D'
)

print('\nDataFrame with Grade column: \n', marks)
```

11. Sort students:

	By marks in ascending order.
	By marks in descending order.
	By Department and then Marks.


```python
print('Students sorted by Marks (Ascending): \n',
      marks.sort_values('Marks'))

print('\nStudents sorted by Marks (Descending): \n',
      marks.sort_values('Marks', ascending=False))

print('\nStudents sorted by Department and then Marks: \n',
      marks.sort_values(['Department', 'Marks']))
```

12. Consider:

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

Find:

Missing values
Number of missing values in each column
Rows containing missing values



```python
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

print('Missing values: \n', students.isnull())

print('\nNumber of missing values in each column: \n',
      students.isnull().sum())

print('\nRows containing missing values: \n',
      students[students.isnull().any(axis=1)])
```
