```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
```

```python
# Various Features of a DataSet
X = pd.DataFrame({
    'constant-feature': [1, 1, 1, 1, 1],
    'almost_constant': [0, 0, 0, 0, 1],
    'usefule_feature': [10, 20, 30, 40, 50]
})

selector = VarianceThreshold(threshold=0.18) # threshold=0.0
X_selected = selector.fit_transform(X)
selected_columns = X.columns[selector.get_support()]

# selected_columns
print('Selected Features : ', list(selected_columns))
```
